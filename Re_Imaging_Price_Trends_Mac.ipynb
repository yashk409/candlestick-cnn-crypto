# 1:1 Replication: (Re-)Imag(in)ing Price Trends (M4 Mac 16GB Optimized)
This notebook is a full replication of the experiments from "(Re-)Imag(in)ing Price Trends" (Jiang, Kelly, Xiu 2023).
**Mac M4 16GB Optimization:** This notebook has been specifically refactored to use "Lazy Loading." Instead of generating 7.5 million images up-front (which would require ~115 GB of RAM and crash your Mac), it stores only lightweight data indices and generates images on-the-fly using the M4's blazing fast CPU cores, feeding them directly to the `mps` GPU engine.


```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm
import time
import os
import requests
import warnings
warnings.filterwarnings('ignore')

# The M4 chip relies on Metal Performance Shaders (MPS) for GPU acceleration.
# device = torch.device('cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu')
# print(f"Using device: {device}")
```

## 3. Image Generation


```python
def generate_image(window_df, n_days):
    if n_days == 5: height = 32
    elif n_days == 20: height = 64
    else: height = 96
    width = n_days * 3
    
    img = Image.new('L', (width, height), color=0)
    draw = ImageDraw.Draw(img)
    
    if len(window_df) == 0: return np.zeros((1, height, width), dtype=np.float32)
        
    first_close = window_df['Close'].iloc[0]
    if first_close == 0 or np.isnan(first_close): return np.zeros((1, height, width), dtype=np.float32)
    
    prices = window_df[['Open', 'High', 'Low', 'Close']].values / first_close
    ma = window_df['Close'].rolling(window=n_days, min_periods=1).mean().values / first_close
    volumes = window_df['Volume'].values
    
    max_p, min_p = np.nanmax(prices), np.nanmin(prices)
    max_v = np.nanmax(volumes)
    if max_p == min_p: max_p = min_p + 1e-5
    if max_v == 0: max_v = 1e-5
    
    price_h = int(height * 0.8)
    vol_h = height - price_h
    
    def price_to_y(p): return int(np.clip(price_h - 1 - (p - min_p) / (max_p - min_p) * (price_h - 1), 0, price_h - 1))
    def vol_to_y(v): return int(np.clip(height - 1 - (v / max_v) * (vol_h - 1), price_h, height - 1))
    
    ma_points = []
    for i in range(len(window_df)):
        x_center = i * 3 + 1
        o_y = price_to_y(prices[i, 0])
        h_y = price_to_y(prices[i, 1])
        l_y = price_to_y(prices[i, 2])
        c_y = price_to_y(prices[i, 3])
        
        draw.line([(x_center, h_y), (x_center, l_y)], fill=255, width=1)
        draw.line([(x_center-1, o_y), (x_center, o_y)], fill=255, width=1)
        draw.line([(x_center, c_y), (x_center+1, c_y)], fill=255, width=1)
        
        v_y = vol_to_y(volumes[i])
        draw.line([(x_center, height-1), (x_center, v_y)], fill=255, width=1)
        
        ma_y = price_to_y(ma[i])
        ma_points.append((x_center, ma_y))
        
    if len(ma_points) > 1:
        draw.line(ma_points, fill=255, width=1)
        
    img_array = np.array(img, dtype=np.float32) / 255.0
    return np.expand_dims(img_array, axis=0)
```

## 4. Mac 16GB Optimized Lazy Dataloader
Instead of holding all generated images in RAM, this Dataset only holds lightweight references and uses the M4 CPU to generate images dynamically when requested by the MPS GPU during training.


```python
class LazyStockImageDataset(Dataset):
    def __init__(self, df, n_days=20, horizon=20):
        self.n_days = n_days
        self.samples = []
        self.grouped_dfs = {}
        
        tickers = df['Ticker'].unique()
        for ticker in tqdm(tickers, desc=f"Indexing Data (N={n_days}, H={horizon})", leave=False):
            t_df = df[df['Ticker'] == ticker].copy()
            t_df = t_df.sort_values('Date').reset_index(drop=True)
            
            # Forward return
            t_df['Target_Ret'] = t_df['Close'].shift(-horizon) / t_df['Close'] - 1
            t_df['Label'] = (t_df['Target_Ret'] > 0).astype(int)
            
            self.grouped_dfs[ticker] = t_df
            
            # Only create lightweight pointers to valid rows to save RAM
            for i in range(len(t_df) - n_days):
                window = t_df.iloc[i:i+n_days]
                target = t_df['Label'].iloc[i+n_days-1]
                
                # Check for NaNs
                if window.isna().any().any() or pd.isna(target):
                    continue
                    
                base = {
                    'Date': str(t_df['Date'].iloc[i+n_days-1].date()),
                    'Ticker': ticker,
                    'MOM': window['MOM'].iloc[-1],
                    'STR': window['STR'].iloc[-1],
                    'WSTR': window['WSTR'].iloc[-1],
                    'TREND': window['TREND'].iloc[-1],
                    'Fwd_Ret': t_df['Target_Ret'].iloc[i+n_days-1]
                }
                
                self.samples.append({
                    'ticker': ticker,
                    'start_idx': i,
                    'label': int(target),
                    'base': base
                })
                
    def __len__(self): return len(self.samples)
    
    def __getitem__(self, idx):
        # Generate the image on the fly!
        sample = self.samples[idx]
        t_df = self.grouped_dfs[sample['ticker']]
        start = sample['start_idx']
        
        window = t_df.iloc[start:start+self.n_days]
        img = generate_image(window, self.n_days)
        
        return torch.tensor(img), torch.tensor(sample['label'], dtype=torch.long), sample['base']

def create_dataloaders(df, n_days, horizon, batch_size=128):
    train_val_df = df[(df['Date'] >= '1997-01-01') & (df['Date'] <= '2004-12-31')]
    test_df = df[(df['Date'] >= '2005-01-01') & (df['Date'] <= '2023-12-31')]
    
    train_val_ds = LazyStockImageDataset(train_val_df, n_days, horizon)
    
    if len(train_val_ds) == 0: return None, None, None
        
    train_size = int(0.7 * len(train_val_ds))
    val_size = len(train_val_ds) - train_size
    train_ds, val_ds = torch.utils.data.random_split(train_val_ds, [train_size, val_size])
    
    test_ds = LazyStockImageDataset(test_df, n_days, horizon)
    
    # Using num_workers=0. Jupyter Notebooks on macOS use the 'spawn' multiprocessing method, 
    # which cannot pickle local classes defined in the notebook's __main__ scope.
    # The M4's single-core CPU performance is incredibly fast, so on-the-fly generation 
    # without multiprocessing will still easily saturate the GPU.
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=0)
    
    return train_loader, val_loader, test_loader
```

## 5. CNN Architecture (PyTorch)


```python
def xavier_init(m):
    if isinstance(m, nn.Conv2d) or isinstance(m, nn.Linear):
        nn.init.xavier_uniform_(m.weight)
        if m.bias is not None:
            nn.init.zeros_(m.bias)

class TrendCNN(nn.Module):
    def __init__(self, n_days=20):
        super(TrendCNN, self).__init__()
        
        if n_days == 5: blocks, v_stride, v_dilation = 2, 1, 1
        elif n_days == 20: blocks, v_stride, v_dilation = 3, 3, 2
        else: blocks, v_stride, v_dilation = 4, 3, 3
            
        layers = []
        in_channels, out_channels = 1, 64
        
        for i in range(blocks):
            stride = (v_stride, 1) if i == 0 else (1, 1)
            dilation = (v_dilation, 1) if i == 0 else (1, 1)
            layers.append(nn.Conv2d(in_channels, out_channels, kernel_size=(5, 3), stride=stride, padding=(2, 1), dilation=dilation))
            layers.append(nn.BatchNorm2d(out_channels))
            layers.append(nn.LeakyReLU(0.01))
            layers.append(nn.MaxPool2d(kernel_size=(2, 1)))
            in_channels = out_channels
            out_channels *= 2
            
        self.features = nn.Sequential(*layers)
        
        test_tensor = torch.zeros(1, 1, 32 if n_days==5 else (64 if n_days==20 else 96), n_days * 3)
        flat_size = self.features(test_tensor).view(1, -1).size(1)
        
        self.classifier = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(flat_size, 2)
        )
        self.apply(xavier_init)
        
    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x
```

## 6. Training & Evaluation Engine


```python
def train_model(model, train_loader, val_loader, epochs=100, lr=1e-5, patience=2):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    model.to(device)
    
    best_loss = float('inf')
    patience_counter = 0
    
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for imgs, labels, _ in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
            
        model.eval()
        val_loss = 0
        with torch.no_grad():
            for imgs, labels, _ in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = criterion(outputs, labels)
                val_loss += loss.item()
                
        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        
        if val_loss < best_loss:
            best_loss = val_loss
            patience_counter = 0
            torch.save(model.state_dict(), 'best_model.pt')
        else:
            patience_counter += 1
            if patience_counter >= patience: break
                
    try: model.load_state_dict(torch.load('best_model.pt', weights_only=True))
    except: model.load_state_dict(torch.load('best_model.pt'))
    return model

def generate_predictions(model, test_loader):
    model.eval()
    model.to(device)
    predictions, baselines_list = [], []
    with torch.no_grad():
        for imgs, _, baselines in test_loader:
            imgs = imgs.to(device)
            probs = torch.softmax(model(imgs), dim=1)[:, 1].cpu().numpy()
            predictions.extend(probs)
            keys = baselines.keys()
            for i in range(len(probs)):
                b = {k: (baselines[k][i].item() if isinstance(baselines[k][i], torch.Tensor) else baselines[k][i]) for k in keys}
                baselines_list.append(b)
                
    df_eval = pd.DataFrame(baselines_list)
    df_eval['CNN_Prob'] = predictions
    return df_eval
```

## 7. Full 1:1 Experimentation Loop


```python
IMAGE_SIZES = [5, 20, 60]
HORIZONS = [5, 20, 60]
ENSEMBLE_SIZE = 5

if TESTING:
    IMAGE_SIZES = [5]
    HORIZONS = [5]
    ENSEMBLE_SIZE = 1
    
results = []

for n_days in IMAGE_SIZES:
    for horizon in HORIZONS:
        print(f"\n{'='*50}\nRunning Configuration: I{n_days} / R{horizon}\n{'='*50}")
        
        train_loader, val_loader, test_loader = create_dataloaders(df_market, n_days, horizon, batch_size=128)
        if train_loader is None or len(train_loader) == 0:
            print("Not enough data to train this configuration. Skipping.")
            continue
            
        all_test_preds = []
        for i in range(ENSEMBLE_SIZE):
            print(f"--- Training Ensemble Model {i+1}/{ENSEMBLE_SIZE} ---")
            model = TrendCNN(n_days=n_days)
            model = train_model(model, train_loader, val_loader, epochs=100, lr=1e-5, patience=2)
            df_eval = generate_predictions(model, test_loader)
            all_test_preds.append(df_eval['CNN_Prob'].values)
            
        avg_preds = np.mean(all_test_preds, axis=0)
        df_eval['CNN_Prob'] = avg_preds
        
        def calc_sharpe(signal_col, df):
            try:
                df['Decile'] = df.groupby('Date')[signal_col].transform(lambda x: pd.qcut(x.rank(method='first'), 10, labels=False))
                high_rets = df[df['Decile'] == 9].groupby('Date')['Fwd_Ret'].mean()
                low_rets = df[df['Decile'] == 0].groupby('Date')['Fwd_Ret'].mean()
                spread_rets = high_rets - low_rets
                
                avg_spread = spread_rets.mean()
                std_spread = spread_rets.std() / np.sqrt(horizon)
                annual_vol = std_spread * np.sqrt(252)
                annual_spread = avg_spread * (252 / horizon)
                
                sharpe = annual_spread / annual_vol if annual_vol != 0 else 0
                return annual_spread, sharpe
            except Exception as e:
                return 0, 0
                
        strategies = ['CNN_Prob', 'MOM', 'STR', 'WSTR', 'TREND']
        for strat in strategies:
            ret, sharpe = calc_sharpe(strat, df_eval.copy())
            results.append({
                'Config': f"I{n_days}/R{horizon}",
                'Strategy': strat,
                'Ann_Return': ret,
                'Ann_Sharpe': sharpe
            })
            
df_results = pd.DataFrame(results)
display(df_results)
df_results.to_csv("final_experiment_results.csv", index=False)
print("\nExperimentation Complete! Results saved to final_experiment_results.csv")
```
